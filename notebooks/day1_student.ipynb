### 1. 내 연구에서 예측하고 싶은 숫자 하나
머신러닝은 결국 **단서를 보고 숫자(또는 분류)를 맞히는 일**입니다. 내 전공에서 미리 알고 싶은 숫자를 하나 떠올려 보세요.
예 — 「사료 배합 → 돼지 체중」 「날씨 · 요일 → 응급실 방문 수」 「예산 · 인구 → 복지 수요」 「공정 온도 → 수율」

✅ **체크포인트 ②** — 채팅에 **「전공 / 맞히고 싶은 숫자 / 단서가 될 것」** 한 줄을 올려 주세요.

이 셀이 하는 일 — 예시 5줄 아래 **마지막 줄을 내 것으로 바꿔** 표로 만듭니다. 따옴표 안의 글자만 바꾸고 따옴표 · 쉼표는 그대로 둡니다. 전공은 달라도 **모양은 전부 같습니다**: 단서 → 숫자.

```python
import pandas as pd

ours = [
    ["수의학",   "돼지 체중(kg)",     "사료 배합 · 일령 · 품종"],
    ["행정학",   "민원 건수(건/월)",   "인구 · 계절 · 정책 시행"],
    ["아동복지", "돌봄 수요(명)",      "맞벌이 비율 · 지역 · 연령"],
    ["반도체",   "수율(%)",           "공정 온도 · 압력 · 시간"],
    ["간호학",   "응급실 방문(명/일)", "요일 · 날씨 · 유행 질환"],
    ["내 전공",  "맞히고 싶은 숫자",   "단서가 될 것"],          # ← 이 줄을 내 것으로
]

pd.DataFrame(ours, columns=["전공", "맞히고 싶은 숫자", "단서"])
```

오늘부터 4일 동안은 **모두에게 같은 데이터**로 이 모양을 연습합니다 — **서울 따릉이 대여 대수**. 단서는 시각 · 요일 · 기온 · 비.
마지막 날에는 같은 노트북에 **내 연구 데이터**를 넣습니다.

### 2. 따릉이 데이터 읽기
이 셀이 하는 일 — 인터넷의 CSV 파일을 **표(DataFrame)** 로 읽어 옵니다. `pandas`는 표를 다루는 도구 모음.
데이터: 2017년 12월 ~ 2018년 10월, 서울시 따릉이 **한 시간 = 1줄**. 그 시간에 대여된 대수와 그때 날씨.

```python
BASE = "https://raw.githubusercontent.com/aicec/seoul-bike-ml-data/main/"
bikes = pd.read_csv(BASE + "bikes_hourly.csv")

bikes.head()     # 앞 5줄 보기
```

열 뜻 — `date` 날짜 · `hour` 시각 · **`rentals` 대여 대수(예측 대상)** · `temp_c` 기온 · `humidity_pct` 습도 · `rain_mm` 강수량 · `day_name` 요일 · `is_weekend` 주말이면 1 · `functioning` 운영한 시간이면 1

## 블록 2 · 파이썬 읽는 법과 하루 곡선

### 3. 파이썬 읽기 ① 변수 · 리스트 · 반복
이 셀이 하는 일 — 어느 날 저녁 6시간의 대여 대수를 **리스트**에 담고, **for**로 하나씩 꺼내 막대로 찍습니다.
- `evening = [...]` 이름표를 붙여 보관 = **변수**
- `[ , , ]` 여러 값을 순서대로 = **리스트**
- `for n in evening:` 리스트에서 하나씩 꺼내 `n`이라고 부름 = **반복**

```python
evening = [1458, 1951, 2915, 2248, 1958, 1944]   # 2018-05-15(화) 16시 ~ 21시 대여 대수
hour = 16

for n in evening:
    print(hour, "시", n, "대 ", "■" * (n // 200))
    hour = hour + 1
```

### 4. 파이썬 읽기 ② 함수 부르기
이 셀이 하는 일 — 이미 만들어진 도구(**함수**)를 부릅니다. `len()` 개수, `max()` 최대, `sum()` 합. 괄호 안에 재료를 넣으면 결과가 나옵니다.

```python
print("몇 시간:", len(evening))
print("가장 많이:", max(evening), " 가장 적게:", min(evening))
print("합계:", sum(evening), " 평균:", round(sum(evening) / len(evening)))
```

### 5. 표의 크기와 기간
이 셀이 하는 일 — 표의 줄 수, 기간, 열 이름을 확인합니다.

```python
print("행(줄) 수:", len(bikes))
print("기간:", bikes.date.min(), "~", bikes.date.max())
print("열:", bikes.columns.tolist())
```

✅ **체크포인트 ③** — `행 수`를 채팅에 올려 주세요.

### 6. 표 다루기 — 하루만 골라내기
이 셀이 하는 일 — `date`가 `"2018-05-15"`인 줄만 골라 `day`라는 새 표를 만듭니다. `표[조건]` 꼴. 24줄이 나와야 합니다.

```python
day = bikes[bikes.date == "2018-05-15"]

print(len(day), "줄 ·", day.day_name.iloc[0])
day[["hour", "rentals", "temp_c", "rain_mm"]].head(10)
```

### 7. 하루 곡선 — 몇 시에 많이 나가나
이 셀이 하는 일 — x축 시각, y축 대여 대수로 **선 그래프**를 그립니다.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))
plt.plot(day.hour, day.rentals, marker="o")
plt.xlabel("hour"); plt.ylabel("rentals"); plt.title("2018-05-15 (Tue)")
plt.xticks(range(0, 24, 2)); plt.grid(alpha=0.3)
plt.show()
```

✅ **체크포인트 ④** — 그래프를 캡처해 PAD 게시판 1일차에 올려 주세요(수업 끝까지 올리면 됩니다).

**눈으로 읽기** — 봉우리가 **몇 개**, **몇 시**에 있나요? 그 시간에 서울에서 무슨 일이 있나요?

## 블록 3 · 평일 · 주말 · 날씨 · 결정

### 8. 평일 vs 주말 — 시간별 평균 겹쳐 그리기
이 셀이 하는 일 — 운영한 시간만 남기고(`functioning == 1`), 주말 여부와 시각으로 묶어 **평균**을 내고, 선 두 개를 겹칩니다.

```python
open_hours = bikes[bikes.functioning == 1]                 # 운영 안 한 시간(대여 0)은 뺌

plt.figure(figsize=(8, 4))
for wk, g in open_hours.groupby("is_weekend"):
    avg = g.groupby("hour").rentals.mean()
    plt.plot(avg.index, avg.values, marker=".", label="weekend" if wk == 1 else "weekday")
plt.xlabel("hour"); plt.ylabel("average rentals"); plt.xticks(range(0, 24, 2))
plt.legend(); plt.grid(alpha=0.3)
plt.show()
```

**읽기** — 평일은 8시 · 18시 두 봉우리, 주말은 오후에 하나. **같은 시각이라도 요일에 따라 다릅니다** → 요일이 단서가 됩니다.

### 9. 기온과 대여량 — 점으로 보기
이 셀이 하는 일 — 한 시간을 점 하나로, x축 기온 · y축 대여 대수 **산점도**를 그립니다. 8,000개 점이라 반투명으로.

```python
plt.figure(figsize=(8, 4))
plt.scatter(open_hours.temp_c, open_hours.rentals, s=4, alpha=0.25)
plt.xlabel("temperature (C)"); plt.ylabel("rentals"); plt.grid(alpha=0.3)
plt.show()
```

**읽기** — 추우면 적고, 더울수록 늘다가, 30 ℃를 넘으면 다시 줄어듭니다. **직선이 아닙니다**(내일 이야기).

### 10. 비가 오면 얼마나 줄까 — 코드 두 줄
이 셀이 하는 일 — 비 오는 시간과 안 오는 시간을 조건으로 나눠 **평균**을 비교합니다.

```python
rainy = open_hours[open_hours.rain_mm > 0]
dry   = open_hours[open_hours.rain_mm == 0]

print("비 오는 시간 평균:", round(rainy.rentals.mean()), "대")
print("안 오는 시간 평균:", round(dry.rentals.mean()), "대")

wd18 = open_hours[(open_hours.hour == 18) & (open_hours.is_weekend == 0)]
print("평일 18시 — 비:", round(wd18[wd18.rain_mm > 0].rentals.mean()), "대 · 맑음:", round(wd18[wd18.rain_mm == 0].rentals.mean()), "대")
```

✅ **체크포인트 ⑤** — `비 오는 시간 평균`을 채팅에 올려 주세요.

### 미션 1 · 다른 날의 하루 곡선

### 미션 2 · 습도도 단서일까
9번 셀의 x축을 기온 대신 **습도**(`humidity_pct`)로 바꿔 그리세요. 기온만큼 뚜렷한가요?

```python
plt.figure(figsize=(8, 4))
plt.scatter(open_hours.temp_c, open_hours.rentals, s=4, alpha=0.25, color="orange")   # ← temp_c 를 humidity_pct 로
plt.xlabel("humidity (%)"); plt.ylabel("rentals"); plt.grid(alpha=0.3); plt.show()
```

### 미션 3 (도전) · 달마다 얼마나 다른가
`groupby("month")` 로 **월별 평균** 대여 대수를 막대로 그리세요. 힌트: `avg = open_hours.groupby("month").rentals.mean()` 뒤에 `plt.bar(avg.index, avg.values)`.

```python
avg = open_hours.groupby("hour").rentals.mean()      # ← "hour" 를 "month" 로

plt.figure(figsize=(8, 4))
plt.bar(avg.index, avg.values)
plt.xlabel("month"); plt.ylabel("average rentals per hour"); plt.xticks(range(1, 13)); plt.grid(alpha=0.3, axis="y")
plt.show()
```
